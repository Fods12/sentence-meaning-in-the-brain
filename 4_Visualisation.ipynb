```python
## Script for plotting surface brain images
# Developed from code here https://nilearn.github.io/stable/auto_examples/01_plotting/plot_3d_map_to_surface_projection.html
# Run with python 3.10 fMRI env
# James Fodor 2024
# Best practise for significance testing using p-values for display is given in
# 'Attention weights accurately predict language representations in the brain'

import numpy as np
import nibabel
import json
import seaborn as sns

from nilearn import surface
from nilearn import plotting
from nibabel import processing

# load custom functions from library
import sentence_embeds_processing as sep
```

### Specify data to plot

```python
## Specify parameters and load brain coordinates

# Specify data to plot
dataset_name = 'Fodor2024-final108'
timing = 'mid' # start, end, mid
model_name = 'parse-GPT4_CN_fixed_control-1_set-all_size-8_logp-tfce-big.nii.gz'
```

```python
## Prepare and produce the images

# Load the nii image
input_path = 'Data-fmri/searchlight/'+timing+'_3s/'+model_name
try:
    image_nii = nibabel.load(input_path)
except FileNotFoundError:
    print('File not found')

# Perform simple nii processing
image_np = image_nii.get_fdata()
image_np_mod = np.nan_to_num(image_np) # remove NaNs
image_np_mod = image_np_mod.clip(min=0) # cut off negative values
image_nii_mod = nibabel.Nifti1Image(image_np_mod, image_nii.affine)
smooth = 6
image_nii_final = processing.smooth_image(image_nii_mod, fwhm=smooth) # smooth image for visualisation

# Produce image
view = plotting.view_img_on_surf(
    image_nii_final,
    surf_mesh="fsaverage",
    darkness=0.7, # this affects the colour of the sulci
    cmap='hot', # hot (for RSA, use 85%), cold_hot, cool, blue_transparent (for ROI, use 85-95% as needed)
    symmetric_cmap=False,
    vmin=0,
    vmax=5,
    colorbar=True,
    colorbar_height=0.7,
    colorbar_fontsize=36,
    threshold='85%') # this is a percentile

# Generate image
print('Model',model_name)
size_factor = 2
view.resize(2400/size_factor,1800/size_factor)
# view.save_as_html('D:\Downloads\\'+dataset_name+'_'+model_name+'_'+analysis_type+'.html')
```
